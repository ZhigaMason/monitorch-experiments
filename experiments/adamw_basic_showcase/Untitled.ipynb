```python
LOGS = [
    'logs/adam_lr3e-2_wd0.05_moments-default.pkl',
    'logs/adam_lr3e-4_wd0.05_moments0.pkl',
    'logs/adam_lr3e-4_wd0.05_moments-default.pkl',
]
```

```python
from monitorch.visualizer import (
    PlayerVisualizer,
    MatplotlibVisualizer,
    PrintVisualizer
)

figs = []
for filename in LOGS:
    viz = PlayerVisualizer(filename, MatplotlibVisualizer())
    viz.playback()
    fig = viz.target.show_fig()
    figs.append(fig)
```

```python
for fig, filename in zip(figs, LOGS):
    filename_fig = 'figs' + filename[:-3][4:] + 'png'
    fig.savefig(filename_fig)
```
